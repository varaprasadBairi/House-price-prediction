```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Step 1: Separate X and y
X = df.drop('SalePrice', axis=1)   # remove target
y = df['SalePrice']                # target (not used in VIF)

# Step 2: Keep only numerical columns
X = X.select_dtypes(include=['int64', 'float64'])

# Step 3: Create VIF dataframe
vif_data = pd.DataFrame()
vif_data['feature'] = X.columns   # column names

# Step 4: Calculate VIF
vif_data['VIF'] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

# Step 5: Show result
print(vif_data)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

lr_model=LinearRegression()
rf_model = RandomForestRegressor()
gb_model = GradientBoostingRegressor()
xg_model=XGBRegressor()
```

```python
lr_model.fit(x_train,y_train)
```

```python
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

rf_model.fit(x_train, y_train)
```

```python
gb_model = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42)

gb_model.fit(x_train, y_train)
```

```python
xg_model = XGBRegressor(
    n_estimators=1200,
    learning_rate=0.03,
    max_depth=4,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=1,
    random_state=42
)
xg_model.fit(x_train,y_train)
```

```python
y_pred_lr=lr_model.predict(x_test)
y_pred_rf = rf_model.predict(x_test)
y_pred_gb = gb_model.predict(x_test)
y_pred_xg=xg_model.predict(x_test)
```

# Tuning for Linear regression model

```python
lr = LinearRegression()

lr_params = {
    'fit_intercept': [True, False],
    'positive': [True, False]
}

lr_grid = GridSearchCV(
    estimator=lr,
    param_grid=lr_params,
    cv=5,
    scoring='r2',
    n_jobs=-1
)

lr_grid.fit(x_train, y_train)

print("Best Params:", lr_grid.best_params_)
print("Best Score:", lr_grid.best_score_)

best_lr = lr_grid.best_estimator_
```

# Tuning for RandomForestRegressor model

```python
rf_params = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
}
```

```python
rf = RandomForestRegressor(random_state=42)

rf_grid = GridSearchCV(
    estimator=rf,
    param_grid=rf_params,
    cv=5,
    scoring='r2',
    n_jobs=-1
)
rf_grid.fit(x_train, y_train)
```

```python
print("Best RF Params:", rf_grid.best_params_)
```

```python
best_rf = rf_grid.best_estimator_
best_rf
```

# Tuning for GradientBoostingRegressor model

```python
gb_params = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 5, 7]
}
```

```python
gb = GradientBoostingRegressor(random_state=42)

gb_grid = GridSearchCV(
    estimator=gb,
    param_grid=gb_params,
    cv=5,
    scoring='r2',
    n_jobs=-1
)

gb_grid.fit(x_train, y_train)
```

```python
best_gb = gb_grid.best_estimator_
```

Abstract
This project focuses on predicting house prices using machine learning techniques. The dataset containing various house features such as location, size, number of rooms, and other attributes was analyzed and processed. Data cleaning and preprocessing were performed using Python libraries like Pandas and NumPy.

Exploratory Data Analysis (EDA) was conducted to understand patterns and relationships between features affecting house prices. A machine learning model was then built and trained on the processed data to predict house prices accurately.

The trained model was integrated into a backend system using FastAPI, and a frontend interface was developed to allow users to input house details and get predicted prices. This project demonstrates a complete pipeline from data processing to deployment.

1. Introduction
Real estate price prediction is an important application of data science. House prices depend on multiple factors such as location, size, amenities, and infrastructure. By analyzing historical housing data, we can build models to predict prices accurately.

In this project, the dataset was cleaned, analyzed, and used to build a predictive model. The model was then deployed through a backend API and a frontend interface for user interaction.
